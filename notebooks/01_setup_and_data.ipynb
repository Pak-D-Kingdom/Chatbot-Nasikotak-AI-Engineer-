# Notebook 01: Setup & Data

Notebook ini bertujuan untuk menginisialisasi database SQLite, melakukan proses seeding (memasukkan data awal) untuk produk dan promosi, serta memverifikasi bahwa file *knowledge base* (Markdown) sudah siap.

```python
import os
import sys
import glob

# Add src to Python path
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))

from src.database import init_db, SessionLocal, Product, Promotion
from datetime import datetime, timedelta
```

## 1. Initialize Database

```python
# Create tables
init_db()
```

## 2. Seed Database

```python
def seed_data():
    db = SessionLocal()
    try:
        # Cek apakah data sudah ada
        if db.query(Product).count() > 0:
            print("Database already seeded.")
            return
        
        products = [
            Product(id="P001", name="Nasi Kotak Minibox", category="minibox", price=17000, minimum_order=20, suitable_for="gathering, office_break, casual_event"),
            Product(id="P002", name="Nasi Kotak Broiler", category="broiler", price=20000, minimum_order=20, suitable_for="meeting, seminar, training, office_event"),
            Product(id="P003", name="Nasi Kotak Broiler Jumbo", category="broiler_jumbo", price=23000, minimum_order=30, suitable_for="meeting, seminar, workshop, gathering, corporate_event"),
            Product(id="P004", name="Nasi Kotak Ayam Kampung", category="ayam_kampung", price=24000, minimum_order=30, suitable_for="gathering, family_event, celebration, special_event"),
            Product(id="P005", name="Nasi Kotak Bebek Mantap", category="bebek", price=27000, minimum_order=30, suitable_for="gathering, corporate_event, special_occasion, premium_event"),
            Product(id="P006", name="Nasi Kotak Gurami", category="gurami", price=27000, minimum_order=30, suitable_for="gathering, corporate_event, special_occasion, premium_event")
        ]
        
        promotions = [
            Promotion(id="PR01", name="Cashback 5% Corporate", discount_type="percentage", discount_value=5.0, start_date=datetime.utcnow(), end_date=datetime.utcnow() + timedelta(days=30), conditions="Minimum pembelian 5 juta")
        ]
        
        db.add_all(products)
        db.add_all(promotions)
        db.commit()
        print("Seeding successful.")
    except Exception as e:
        print(f"Error seeding: {e}")
        db.rollback()
    finally:
        db.close()

seed_data()
```

## 3. Verify Database

```python
db = SessionLocal()
prods = db.query(Product).all()
print(f"--- {len(prods)} Products found ---")
for p in prods:
    print(f"- {p.id}: {p.name} (Rp {p.price}) [{p.category}]")

promos = db.query(Promotion).all()
print(f"\n--- {len(promos)} Promotions found ---")
for p in promos:
    print(f"- {p.id}: {p.name} (Diskon: {p.discount_value}%)")
db.close()
```
